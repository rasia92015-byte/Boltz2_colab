```python
# @title
# ============================================================
# 1. EXCEL INPUT
# ============================================================

EXCEL_PATH = "/content/PDBbind_Boltz_sequence.xlsx"

# Sheet name:
# 0 = first sheet
# or use actual name, e.g. "Sheet1"
SHEET_NAME = 0


# ============================================================
# 2. EXCEL FIELD NAMES
# ============================================================

JOB_ID_FIELD = "Binding_ID"
SEQUENCE_FIELD = "sequence"
SMILES_FIELD = "smiles"


# ============================================================
# 3. START / END JOB
# ============================================================

# Inclusive:
# START_JOB_ID and END_JOB_ID themselves WILL be run.

START_JOB_ID = "PDBB_00061"
END_JOB_ID   = "PDBB_00120"

# To run from the first job:
# START_JOB_ID = None

# To run through the last job:
# END_JOB_ID = None


# ============================================================
# 4. GOOGLE DRIVE OUTPUT
# ============================================================

DRIVE_ROOT = "/content/drive/MyDrive/Boltz2_Results"

# Final Boltz output:
OUTPUT_DIR = f"{DRIVE_ROOT}/boltz_output"

# Batch logs:
LOG_DIR = f"{DRIVE_ROOT}/logs"

# Final summary Excel:
SUMMARY_XLSX = f"{DRIVE_ROOT}/Boltz2_results.xlsx"


# ============================================================
# 5. BOLTZ
# ============================================================

BOLTZ = "/content/boltz_env/bin/boltz"


# ============================================================
# 6. STRUCTURE PARAMETERS
# ============================================================

RECYCLING_STEPS = 5

SAMPLING_STEPS = 250

DIFFUSION_SAMPLES = 7

STEP_SCALE = 1.438


# ============================================================
# 7. MSA PARAMETERS
# ============================================================

USE_MSA_SERVER = True

MAX_MSA_SEQUENCES = 8192

MSA_PAIRING_STRATEGY = "complete"


# ============================================================
# 8. AFFINITY PARAMETERS
# ============================================================

SAMPLING_STEPS_AFFINITY = 200

DIFFUSION_SAMPLES_AFFINITY = 5


# ============================================================
# 9. GPU
# ============================================================

ACCELERATOR = "gpu"

DEVICES = 1


# ============================================================
# 10. KERNELS
# ============================================================

# Your RTX PRO 6000 Blackwell has already successfully run
# without --no_kernels, so keep this False.
NO_KERNELS = False


# ============================================================
# 11. EXISTING RESULTS
# ============================================================

# False:
#   Keep existing finished predictions and skip them.
#
# True:
#   Re-run and overwrite existing predictions.
#
OVERRIDE = False
# ============================================================
# 11. Re-run existing finished jobs?
# ============================================================
#
# False = finished jobs automatically skipped
# True  = re-run selected jobs

FORCE_RERUN = False


# ============================================================
# 12. Output format
# ============================================================

OUTPUT_FORMAT = "pdb"
```

```python
# @title
import pandas as pd
import os
from pathlib import Path


# ============================================================
# Read Excel
# ============================================================

df = pd.read_excel(
    EXCEL_PATH,
    sheet_name=SHEET_NAME,
    dtype=str
)

print("Total Excel rows:", len(df))
print()
print("Excel columns:")
print(df.columns.tolist())


# ============================================================
# Check required columns
# ============================================================

required_fields = [
    JOB_ID_FIELD,
    SEQUENCE_FIELD,
    SMILES_FIELD
]

missing_fields = [
    x for x in required_fields
    if x not in df.columns
]

if missing_fields:
    raise ValueError(
        f"Missing Excel column(s): {missing_fields}"
    )


# Keep Excel row number
df["_excel_row"] = range(2, len(df) + 2)


# ============================================================
# Clean Job IDs
# ============================================================

df[JOB_ID_FIELD] = (
    df[JOB_ID_FIELD]
    .astype("string")
    .str.strip()
)


# Missing Job ID
missing_job = (
    df[JOB_ID_FIELD].isna()
    | (df[JOB_ID_FIELD] == "")
)

if missing_job.any():

    bad_rows = df.loc[
        missing_job,
        "_excel_row"
    ].tolist()

    raise ValueError(
        f"Missing Job ID at Excel rows: {bad_rows}"
    )


# ============================================================
# Job ID must be unique
# ============================================================

duplicates = df[
    df[JOB_ID_FIELD].duplicated(
        keep=False
    )
]

if len(duplicates) > 0:

    display(
        duplicates[
            [
                "_excel_row",
                JOB_ID_FIELD
            ]
        ]
    )

    raise ValueError(
        "Duplicate Job IDs found. "
        "Every Job ID must be unique."
    )


# ============================================================
# Job ID becomes folder name
# ============================================================

bad_ids = df[
    df[JOB_ID_FIELD].str.contains(
        r"[\\/]",
        regex=True,
        na=False
    )
]

if len(bad_ids) > 0:

    display(
        bad_ids[
            [
                "_excel_row",
                JOB_ID_FIELD
            ]
        ]
    )

    raise ValueError(
        "Job ID cannot contain / or \\ "
        "because Job ID is used as folder name."
    )


# ============================================================
# Find starting Job
# ============================================================

if START_JOB_ID is None:

    start_pos = 0

else:

    matches = df.index[
        df[JOB_ID_FIELD]
        == str(START_JOB_ID)
    ].tolist()

    if not matches:
        raise ValueError(
            f"START_JOB_ID not found: "
            f"{START_JOB_ID}"
        )

    start_pos = matches[0]


# ============================================================
# Find ending Job
# ============================================================

if END_JOB_ID is None:

    end_pos = len(df) - 1

else:

    matches = df.index[
        df[JOB_ID_FIELD]
        == str(END_JOB_ID)
    ].tolist()

    if not matches:
        raise ValueError(
            f"END_JOB_ID not found: "
            f"{END_JOB_ID}"
        )

    end_pos = matches[0]


if end_pos < start_pos:
    raise ValueError(
        "END_JOB_ID occurs before START_JOB_ID "
        "in Excel."
    )


# ============================================================
# Select batch
# ============================================================

batch_df = (
    df.loc[start_pos:end_pos]
    .copy()
    .reset_index(drop=True)
)


print()
print("=" * 60)
print("BATCH SELECTED")
print("=" * 60)

print(
    "Starting Job ID:",
    batch_df.iloc[0][JOB_ID_FIELD]
)

print(
    "Ending Job ID:  ",
    batch_df.iloc[-1][JOB_ID_FIELD]
)

print(
    "Number of jobs: ",
    len(batch_df)
)

print("=" * 60)


display(
    batch_df[
        [
            "_excel_row",
            JOB_ID_FIELD,
            SEQUENCE_FIELD,
            SMILES_FIELD
        ]
    ].head(20)
)
```

Cell 4 — Validate sequences / SMILES

```python
# @title
import pandas as pd


VALID_AA = set(
    "ACDEFGHIKLMNPQRSTVWY"
    "BXZJUO"
)


def clean_sequence(sequence):

    return (
        str(sequence)
        .replace(" ", "")
        .replace("\n", "")
        .replace("\r", "")
        .replace("\t", "")
        .upper()
    )


errors = []


for _, row in batch_df.iterrows():

    job_id = str(
        row[JOB_ID_FIELD]
    ).strip()


    # ========================================================
    # Sequence
    # ========================================================

    sequence = row[SEQUENCE_FIELD]

    if pd.isna(sequence):

        errors.append(
            (
                job_id,
                "Sequence missing"
            )
        )

        continue


    sequence = clean_sequence(sequence)


    if len(sequence) == 0:

        errors.append(
            (
                job_id,
                "Sequence empty"
            )
        )


    invalid = sorted(
        set(sequence) - VALID_AA
    )


    if invalid:

        errors.append(
            (
                job_id,
                f"Invalid amino acid characters: {invalid}"
            )
        )


    # ========================================================
    # SMILES
    # ========================================================

    smiles = row[SMILES_FIELD]

    if pd.isna(smiles):

        errors.append(
            (
                job_id,
                "SMILES missing"
            )
        )

        continue


    smiles = str(smiles).strip()


    if len(smiles) == 0:

        errors.append(
            (
                job_id,
                "SMILES empty"
            )
        )


# ============================================================
# Report
# ============================================================

if errors:

    print("Input errors:")

    for x in errors[:100]:
        print(x)

    raise ValueError(
        f"{len(errors)} input error(s) found."
    )

else:

    print(
        f"✓ All {len(batch_df)} jobs "
        "passed basic validation."
    )
```

Cell 6 — Batch functions

```python
# @title
import os
import json
import shutil
import subprocess
from pathlib import Path
from datetime import datetime


# ============================================================
# YAML-safe quoting
# ============================================================

def yaml_quote(text):
    """
    YAML single quote is escaped by doubling it.
    """
    return str(text).replace("'", "''")


# ============================================================
# Generate YAML
# ============================================================

def generate_yaml(
    job_id,
    sequence,
    smiles
):

    sequence = clean_sequence(sequence)
    smiles = str(smiles).strip()

    return f"""\
version: 1

sequences:
  - protein:
      id: A
      sequence: {sequence}

  - ligand:
      id: B
      smiles: '{yaml_quote(smiles)}'

properties:
  - affinity:
      binder: B
"""


# ============================================================
# Google Drive paths
# ============================================================

def get_job_paths(job_id):

    job_dir = (
        Path(DRIVE_ROOT)
        / job_id
    )

    yaml_path = (
        job_dir
        / f"{job_id}.yaml"
    )

    prediction_dir = (
        job_dir
        / "predictions"
        / job_id
    )

    affinity_file = (
        prediction_dir
        / f"affinity_{job_id}.json"
    )

    pdb_file = (
        prediction_dir
        / f"{job_id}_model_0.pdb"
    )

    confidence_file = (
        prediction_dir
        / f"confidence_{job_id}_model_0.json"
    )

    return {
        "job_dir": job_dir,
        "yaml": yaml_path,
        "prediction_dir": prediction_dir,
        "affinity": affinity_file,
        "pdb": pdb_file,
        "confidence": confidence_file
    }


# ============================================================
# Check whether job is already completed ON GOOGLE DRIVE
# ============================================================

def job_completed(job_id):

    paths = get_job_paths(job_id)

    return (
        paths["affinity"].exists()
        and
        paths["pdb"].exists()
    )


# ============================================================
# Run ONE job
# ============================================================
def run_one_job(job_id, sequence, smiles):

    # ========================================================
    # Drive paths
    # ========================================================

    drive_job_dir = Path(DRIVE_ROOT) / job_id

    drive_yaml = (
        drive_job_dir
        / f"{job_id}.yaml"
    )

    drive_predictions = (
        drive_job_dir
        / "predictions"
    )

    drive_affinity = (
        drive_predictions
        / job_id
        / f"affinity_{job_id}.json"
    )

    drive_pdb = (
        drive_predictions
        / job_id
        / f"{job_id}_model_0.pdb"
    )


    # ========================================================
    # Resume / skip completed job
    # ========================================================

    if (
        drive_affinity.exists()
        and drive_pdb.exists()
        and not FORCE_RERUN
    ):

        print(
            f"✓ {job_id}: already completed on Google Drive -> SKIP"
        )

        return {
            "job_id": job_id,
            "status": "skipped_completed",
            "return_code": 0
        }


    # ========================================================
    # Local job folder
    # ========================================================

    local_job_dir = (
        Path("/content/boltz_batch_work")
        / job_id
    )


    # Clean previous local attempt
    if local_job_dir.exists():

        shutil.rmtree(
            local_job_dir
        )


    local_job_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    # ========================================================
    # Generate YAML
    # ========================================================

    yaml_content = generate_yaml(
        job_id,
        sequence,
        smiles
    )


    local_yaml = (
        local_job_dir
        / f"{job_id}.yaml"
    )


    with open(
        local_yaml,
        "w"
    ) as f:

        f.write(
            yaml_content
        )


    # ========================================================
    # Boltz command
    # ========================================================

    cmd = [
        BOLTZ,
        "predict",
        str(local_yaml),

        "--out_dir",
        str(local_job_dir),

        "--accelerator",
        ACCELERATOR,

        "--devices",
        str(DEVICES),

        "--output_format",
        OUTPUT_FORMAT,

        "--recycling_steps",
        str(RECYCLING_STEPS),

        "--sampling_steps",
        str(SAMPLING_STEPS),

        "--diffusion_samples",
        str(DIFFUSION_SAMPLES),

        "--step_scale",
        str(STEP_SCALE),

        "--max_msa_seqs",
        str(MAX_MSA_SEQUENCES),

        "--msa_pairing_strategy",
        MSA_PAIRING_STRATEGY,

        "--sampling_steps_affinity",
        str(SAMPLING_STEPS_AFFINITY),

        "--diffusion_samples_affinity",
        str(DIFFUSION_SAMPLES_AFFINITY),
    ]


    if USE_MSA_SERVER:
        cmd.append(
            "--use_msa_server"
        )


    if NO_KERNELS:
        cmd.append(
            "--no_kernels"
        )


    # ========================================================
    # Log
    # ========================================================

    log_path = (
        Path(LOG_ROOT)
        / f"{job_id}.log"
    )


    print()
    print("=" * 70)
    print(f"RUNNING: {job_id}")
    print("=" * 70)

    print("Local folder:")
    print(local_job_dir)

    print()

    print("Google Drive destination:")
    print(drive_job_dir)

    print()

    print("Command:")
    print(" ".join(cmd))

    print()


    # ========================================================
    # Run Boltz
    # ========================================================

    with open(
        log_path,
        "w"
    ) as log_file:

        log_file.write(
            f"Job ID: {job_id}\n"
        )

        log_file.write(
            f"Started: {datetime.now()}\n"
        )

        log_file.write(
            "Command:\n"
        )

        log_file.write(
            " ".join(cmd)
            + "\n\n"
        )

        log_file.flush()


        process = subprocess.Popen(
            cmd,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1
        )


        for line in process.stdout:

            print(
                line,
                end=""
            )

            log_file.write(
                line
            )

            log_file.flush()


        return_code = process.wait()


    # ========================================================
    # Boltz failed
    # ========================================================

    if return_code != 0:

        print()
        print(
            f"✘ {job_id}: FAILED "
            f"(return code {return_code})"
        )

        return {
            "job_id": job_id,
            "status": "failed",
            "return_code": return_code
        }


    # ========================================================
    # Find actual Boltz output
    #
    # Current Boltz creates:
    #
    # job_id/
    # └── boltz_results_job_id/
    #     └── predictions/
    # ========================================================

    local_predictions = (
        local_job_dir
        / f"boltz_results_{job_id}"
        / "predictions"
    )


    # Fallback in case future Boltz version changes structure
    if not local_predictions.exists():

        alternative = (
            local_job_dir
            / "predictions"
        )

        if alternative.exists():

            local_predictions = alternative


    # ========================================================
    # Check output
    # ========================================================

    local_affinity = (
        local_predictions
        / job_id
        / f"affinity_{job_id}.json"
    )


    local_pdb = (
        local_predictions
        / job_id
        / f"{job_id}_model_0.pdb"
    )


    print()
    print("Checking local result...")

    print(
        "Predictions folder:",
        local_predictions.exists()
    )

    print(
        "Affinity:",
        local_affinity.exists()
    )

    print(
        "PDB model 0:",
        local_pdb.exists()
    )


    if not (
        local_predictions.exists()
        and local_affinity.exists()
        and local_pdb.exists()
    ):

        print()

        print(
            f"⚠ {job_id}: Boltz finished, "
            "but expected output was not found."
        )

        return {
            "job_id": job_id,
            "status": "incomplete_output",
            "return_code": return_code
        }


    # ========================================================
    # COPY TO GOOGLE DRIVE
    #
    # IMPORTANT:
    # We intentionally remove the
    # boltz_results_JOBID layer.
    # ========================================================

    print()
    print(
        f"Copying {job_id} to Google Drive..."
    )


    # If force rerun, remove previous job folder
    if (
        FORCE_RERUN
        and drive_job_dir.exists()
    ):

        shutil.rmtree(
            drive_job_dir
        )


    drive_job_dir.mkdir(
        parents=True,
        exist_ok=True
    )


    # --------------------------------------------------------
    # Copy YAML
    # --------------------------------------------------------

    shutil.copy2(
        local_yaml,
        drive_yaml
    )


    # --------------------------------------------------------
    # Copy ONLY predictions
    #
    # local:
    # boltz_results_JOBID/predictions
    #
    # Drive:
    # JOBID/predictions
    #
    # --------------------------------------------------------

    shutil.copytree(
        local_predictions,
        drive_predictions,
        dirs_exist_ok=True
    )


    # ========================================================
    # Verify Drive copy
    # ========================================================

    drive_ok = (
        drive_affinity.exists()
        and drive_pdb.exists()
    )


    if drive_ok:

        print()
        print(
            f"✓ {job_id}: SAVED TO GOOGLE DRIVE"
        )

        print(
            drive_job_dir
        )

        status = "completed"


    else:

        print()
        print(
            f"✘ {job_id}: GOOGLE DRIVE COPY FAILED"
        )

        status = "drive_copy_failed"


    # ========================================================
    # Delete local job ONLY after Drive copy succeeds
    # ========================================================

    if drive_ok:

        shutil.rmtree(
            local_job_dir,
            ignore_errors=True
        )

        print(
            f"✓ Local temporary folder removed."
        )


    return {
        "job_id": job_id,
        "status": status,
        "return_code": return_code
    }
```

Cell 8 — RUN THE BATCH

```python
# @title
batch_run_status = []


for i, row in batch_df.iterrows():

    job_id = str(
        row[JOB_ID_FIELD]
    ).strip()


    sequence = row[
        SEQUENCE_FIELD
    ]


    smiles = row[
        SMILES_FIELD
    ]


    print()
    print(
        f"[{i + 1}/{len(batch_df)}] "
        f"{job_id}"
    )


    try:

        result = run_one_job(
            job_id=job_id,
            sequence=sequence,
            smiles=smiles
        )


    except Exception as e:

        print(
            f"✘ {job_id}: Python exception:"
        )

        print(e)


        result = {
            "job_id": job_id,
            "status": f"exception: {e}",
            "return_code": None
        }


    batch_run_status.append(
        result
    )


print()
print("=" * 70)
print("BATCH LOOP FINISHED")
print("=" * 70)
```
