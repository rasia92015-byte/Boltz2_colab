```python
import pandas as pd

# Input files
drug_file = "2B_GPCR_Ligand_Master_SMILES.xlsx"
target_file = "GP_full_sequences.xlsx"

# Save as Excel with sheet name Binding
output_file = "GP_SMILES_full_sequences.xlsx"

prefix = "GP"

# Read Excel files
drug = pd.read_excel(drug_file)
target = pd.read_excel(target_file)

# Join two tables by UniProt_ID
binding = drug.merge(
    target,
    on="UniProt_ID",
    how="inner",
    suffixes=("", "_target")
)

# Optional: remove exact duplicate rows before assigning Binding_ID
binding = binding.drop_duplicates().reset_index(drop=True)

# Check required sorting columns
required_cols = ["Gene", "UniProt_ID", "Ligand_Name"]
missing = [col for col in required_cols if col not in binding.columns]

if missing:
    raise ValueError(f"Missing required columns after join: {missing}")

# Order by Gene, UniProt_ID1, Ligand_name
binding = binding.sort_values(
    by=["Gene", "UniProt_ID", "Ligand_Name"],
    ascending=True
).reset_index(drop=True)

# # Generate Binding_ID: B0001, B0002, ...
# binding.insert(
#     0,
#     "Binding_ID",
#     [f"{prefix}{i:04d}" for i in range(1, len(binding) + 1)]
# )



with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    binding.to_excel(writer, sheet_name="Binding", index=False)

print(f"Saved {len(binding)} rows to {output_file}")
```
