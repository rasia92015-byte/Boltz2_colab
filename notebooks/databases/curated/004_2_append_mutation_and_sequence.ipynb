```python
import pandas as pd

# Input files
file1 = "GP_SMILES_full_sequences.xlsx"   # first table
file2 = "GP_final_mut.xlsx"   # second table


# Output file name
output_file = "GP_SMILES_mut_full_sequences_2_boltz.xlsx"

# Read the only sheet from each Excel file
df1 = pd.read_excel(file1)
df2 = pd.read_excel(file2)

# For 1st table only
df1["Protein_change"] = "_WT"
df1["sequence_4_boltz"] = df1["domain"]

# Append rows, keeping all columns from both files
combined = pd.concat([df1, df2], ignore_index=True, sort=False)

# Save result
combined.to_excel(output_file, index=False)

print(f"Done! Saved as {output_file}")
```
