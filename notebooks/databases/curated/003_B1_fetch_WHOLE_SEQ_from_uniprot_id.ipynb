```python
import time
from pathlib import Path

import requests
from openpyxl import load_workbook


INPUT_XLSX = Path("GP_UniProt_ID.xlsx")
OUTPUT_XLSX = Path("GP_full_sequences.xlsx")

INPUT_SHEET = "Sheet1"
OUTPUT_SHEET = "Full_sequences"


def find_uniprot_id_column(ws):
    """
    Find the UniProt ID column from the first row.

    If no matching header is found, use the first column.
    """
    accepted_headers = {
        "uniprot_id",
        "uniprot id",
        "accession",
        "uniprot",
    }

    for column_index, cell in enumerate(ws[1], start=1):
        if cell.value is None:
            continue

        header = str(cell.value).strip().lower()

        if header in accepted_headers:
            return column_index

    # Fallback to column A
    return 1


def get_full_sequence_from_uniprot(accession, session=None):
    """
    Retrieve the complete canonical protein sequence from UniProt.

    Returns
    -------
    dict
        domain_start = 1
        domain_end   = full protein length
        domain       = complete protein sequence
    """
    accession = str(accession).strip()
    url = f"https://rest.uniprot.org/uniprotkb/{accession}.json"

    if session is None:
        session = requests.Session()

    response = session.get(url, timeout=30)
    response.raise_for_status()

    data = response.json()

    sequence_data = data.get("sequence", {})
    sequence = sequence_data.get("value", "")

    if not sequence:
        raise ValueError(
            f"No protein sequence was returned for {accession}"
        )

    # Remove any unexpected whitespace
    sequence = "".join(sequence.split())

    return {
        "domain_start": 1,
        "domain_end": len(sequence),
        "domain": sequence,
        "status": "OK",
    }


def main():
    if not INPUT_XLSX.exists():
        raise FileNotFoundError(
            f"Input file was not found: {INPUT_XLSX}"
        )

    workbook = load_workbook(INPUT_XLSX)

    if INPUT_SHEET in workbook.sheetnames:
        input_ws = workbook[INPUT_SHEET]
    else:
        input_ws = workbook.active

    id_column = find_uniprot_id_column(input_ws)

    accessions = []

    for row_number in range(2, input_ws.max_row + 1):
        value = input_ws.cell(
            row=row_number,
            column=id_column,
        ).value

        if value is None:
            continue

        accession = str(value).strip()

        if accession:
            accessions.append(accession)

    if OUTPUT_SHEET in workbook.sheetnames:
        del workbook[OUTPUT_SHEET]

    output_ws = workbook.create_sheet(OUTPUT_SHEET)

    output_ws.append([
        "UniProt_ID",
        "domain_start",
        "domain_end",
        "domain",
        "status",
    ])

    session = requests.Session()

    session.headers.update({
        "User-Agent": (
            "UniProt-full-sequence-fetcher/1.0 "
            "(Python requests)"
        )
    })

    # Cache results so duplicate UniProt IDs are fetched only once
    sequence_cache = {}

    total = len(accessions)

    for index, accession in enumerate(accessions, start=1):
        print(
            f"[{index}/{total}] Fetching full sequence "
            f"for {accession} ..."
        )

        try:
            if accession not in sequence_cache:
                sequence_cache[accession] = (
                    get_full_sequence_from_uniprot(
                        accession=accession,
                        session=session,
                    )
                )

                # Be polite to the UniProt API
                time.sleep(0.2)

            result = sequence_cache[accession]

            output_ws.append([
                accession,
                result["domain_start"],
                result["domain_end"],
                result["domain"],
                result["status"],
            ])

        except Exception as error:
            output_ws.append([
                accession,
                "",
                "",
                "",
                (
                    f"ERROR: {type(error).__name__}: "
                    f"{error}"
                ),
            ])

    # Header formatting
    for cell in output_ws[1]:
        cell.style = "Headline 3"

    output_ws.column_dimensions["A"].width = 16
    output_ws.column_dimensions["B"].width = 14
    output_ws.column_dimensions["C"].width = 14
    output_ws.column_dimensions["D"].width = 100
    output_ws.column_dimensions["E"].width = 50

    output_ws.freeze_panes = "A2"
    output_ws.auto_filter.ref = output_ws.dimensions

    workbook.save(OUTPUT_XLSX)

    successful = sum(
        1
        for result in sequence_cache.values()
        if result.get("status") == "OK"
    )

    print()
    print(f"Done: {OUTPUT_XLSX}")
    print(f"Input rows: {total}")
    print(f"Unique UniProt IDs: {len(sequence_cache)}")
    print(f"Successful sequences: {successful}")


if __name__ == "__main__":
    main()
```

```python
import time
import requests
from pathlib import Path
from openpyxl import load_workbook

INPUT_XLSX = Path("Untitled spreadsheet.xlsx")
OUTPUT_XLSX = Path("Untitled spreadsheet_with_kinase_domains.xlsx")

# To exactly match your original function, use:
# MATCH_TERMS = ("kinase",)
# This broader set also captures lipid-kinase domains labeled PI3K/PI4K by UniProt.
MATCH_TERMS = ("kinase", "pi3k", "pi4k")

def find_uniprot_id_column(ws):
    headers = [cell.value for cell in ws[1]]
    for idx, value in enumerate(headers, start=1):
        if value and str(value).strip().lower() in {"uniprot_id", "uniprot id", "accession", "uniprot"}:
            return idx
    # Fallback: first column
    return 1

def get_kinase_domains_from_uniprot(accession):
    url = f"https://rest.uniprot.org/uniprotkb/{accession}.json"
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    data = response.json()

    seq = data["sequence"]["value"]
    domains = []

    for feat in data.get("features", []):
        if feat.get("type") != "Domain":
            continue

        desc = feat.get("description", "")
        desc_l = desc.lower()

        if any(term in desc_l for term in MATCH_TERMS):
            start = int(feat["location"]["start"]["value"])
            end = int(feat["location"]["end"]["value"])

            domain_start = max(1, start - 10)
            domain_end = min(len(seq), end + 10)

            domains.append({
                "description": desc,
                "start": start,
                "end": end,
                "sequence": seq[start - 1:end],
                "domain_start": domain_start,
                "domain_end": domain_end,
                "domain": seq[domain_start - 1:domain_end],
                "source_url": url,
                "status": "OK",
            })

    if not domains:
        domains.append({
            "description": "",
            "start": "",
            "end": "",
            "sequence": "",
            "domain_start": "",
            "domain_end": "",
            "domain": "",
            "source_url": url,
            "status": "No matching kinase Domain feature found",
        })

    return domains

def main():
    wb = load_workbook(INPUT_XLSX)
    input_ws = wb["Sheet1"] if "Sheet1" in wb.sheetnames else wb.active
    id_col = find_uniprot_id_column(input_ws)

    accessions = []
    for row in range(2, input_ws.max_row + 1):
        value = input_ws.cell(row=row, column=id_col).value
        if value:
            accessions.append(str(value).strip())

    if "Kinase_domains" in wb.sheetnames:
        del wb["Kinase_domains"]

    out_ws = wb.create_sheet("Kinase_domains")
    out_ws.append([
        "UniProt_ID",
        "domain_description",
        "start",
        "end",
        "sequence",
        "domain_start",
        "domain_end",
        "domain_start_minus_10_to_end_plus_10",
        "source_url",
        "status",
    ])

    for accession in accessions:
        print(f"Fetching {accession} ...")
        try:
            domains = get_kinase_domains_from_uniprot(accession)
            for d in domains:
                out_ws.append([
                    accession,
                    d["description"],
                    d["start"],
                    d["end"],
                    d["sequence"],
                    d["domain_start"],
                    d["domain_end"],
                    d["domain"],
                    d["source_url"],
                    d["status"],
                ])
        except Exception as e:
            out_ws.append([
                accession, "", "", "", "", "", "",
                "",
                f"https://rest.uniprot.org/uniprotkb/{accession}.json",
                f"ERROR: {type(e).__name__}: {e}",
            ])

        # Be polite to the API.
        time.sleep(0.2)

    # Basic formatting
    for col in range(1, out_ws.max_column + 1):
        out_ws.cell(row=1, column=col).style = "Headline 3"

    widths = {
        "A": 14, "B": 24, "C": 10, "D": 10, "E": 60,
        "F": 12, "G": 12, "H": 70, "I": 50, "J": 42,
    }
    for col, width in widths.items():
        out_ws.column_dimensions[col].width = width

    out_ws.freeze_panes = "A2"
    wb.save(OUTPUT_XLSX)
    print(f"Done: {OUTPUT_XLSX}")

if __name__ == "__main__":
    main()

```
