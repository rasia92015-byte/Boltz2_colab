```python
import pandas as pd
from pathlib import Path

# Current folder, not MyDrive
folder = Path.cwd()

binding_file = folder / "PDBbind_Boltz_sequence.xlsx"
boltz_file = folder / "Boltz2_PDB_summary.xlsx"

out_file = folder / "PDBbind_E_B_affinity.xlsx"

# Read files
binding = pd.read_excel(binding_file, dtype=str)
boltz = pd.read_excel(boltz_file, dtype=str)

# Clean column names
binding.columns = binding.columns.str.strip()
boltz.columns = boltz.columns.str.strip()

# Clean keys
binding["Binding_ID"] = binding["Binding_ID"].astype(str).str.strip()
boltz["jobname"] = boltz["jobname"].astype(str).str.strip()

# Optional: remove duplicate jobname rows in Boltz2 summary
boltz = boltz.drop_duplicates(subset=["jobname"], keep="first")

# Join:
# binding.Binding_ID  =  boltz.jobname
merged = binding.merge(
    boltz,
    left_on="Binding_ID",
    right_on="jobname",
    how="left"
)

# Optional: drop duplicated key column from Boltz2_summary
merged = merged.drop(columns=["jobname"])

# Add match status
boltz_result_cols = [
    "affinity_pred_value",
    "Converted_IC50_nM",
    "affinity_probability_binary",
    "confidence_score",
    "complex_plddt"
]

existing_result_cols = [c for c in boltz_result_cols if c in merged.columns]

merged["Boltz2_match_status"] = merged[existing_result_cols].notna().any(axis=1).map(
    {True: "matched", False: "no Boltz2 result"}
)

# Save
merged.to_excel(out_file, index=False)

print(f"Saved joined file to: {out_file}")
print(f"Binding rows: {len(binding)}")
print(f"Boltz2 rows: {len(boltz)}")
print(f"Joined rows: {len(merged)}")
print(merged["Boltz2_match_status"].value_counts())

merged.head()
```
